# 🧹 Johansen Output Cleanup Utility

This notebook helps you bulk-delete specific files across all simulation run folders in `well_csvs/` to save disk space.
You can run it in **Dry Run** mode first to see exactly what will be deleted and how much space you'll save.

```python
import os
import glob

WELL_CSVS_DIR = "/Users/apple/Desktop/study/programming/Matlab/Plugins/MRST-2026a/core/examples/data/Johansen/well_csvs/"

# ─── 1. Specify the files you want to delete ─────────────────────────────────
# You can use exact names or wildcards (e.g., '*.png', 'saturation_*.png')
FILES_TO_DELETE = [
    "well_locations.png",
    "well_performance.png",
    # "*.png",  # Uncomment to delete all images
    # "saturation_xsec_*.png"
]

print(f"Target directory: {WELL_CSVS_DIR}")
print(f"Target files to delete: {FILES_TO_DELETE}")
```

```python
# ─── 2. DRY RUN (Preview what will be deleted) ─────────────────────────────

total_bytes_to_free = 0
files_found = []

# Find all subdirectories in well_csvs/
run_folders = [d for d in glob.glob(os.path.join(WELL_CSVS_DIR, '*')) if os.path.isdir(d)]

for folder in run_folders:
    for pattern in FILES_TO_DELETE:
        search_path = os.path.join(folder, pattern)
        for file_path in glob.glob(search_path):
            if os.path.isfile(file_path):
                size = os.path.getsize(file_path)
                total_bytes_to_free += size
                files_found.append(file_path)

print(f"🔍 DRY RUN RESULTS:")
print(f"Found {len(files_found)} files matching your criteria across {len(run_folders)} folders.")
print(f"Estimated space to free: {total_bytes_to_free / (1024*1024):.2f} MB")
if len(files_found) > 0:
    print("\nFirst 5 files to be deleted:")
    for f in files_found[:5]:
        print(f"  - {f}")
```

```python
# ─── 3. EXECUTE DELETION (⚠️ This cannot be undone!) ───────────────────────

deleted_count = 0
freed_bytes = 0

for file_path in files_found:
    try:
        size = os.path.getsize(file_path)
        os.remove(file_path)
        freed_bytes += size
        deleted_count += 1
    except Exception as e:
        print(f"❌ Failed to delete {file_path}: {e}")

print("✅ CLEANUP COMPLETE!")
print(f"Successfully deleted {deleted_count} files.")
print(f"Freed up {freed_bytes / (1024*1024):.2f} MB of disk space.")
```
